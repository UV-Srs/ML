```python
import pandas as pd

df1 = pd.read_csv("../data/raw/supplier_delivery_data_90000.csv")
df2 = pd.read_csv("../data/raw/supplier_delivery_data_150000_synthetic.csv")

print("Dataset 1:", df1.shape)
print("Dataset 2:", df2.shape)

# Check matching columns
print("Same columns:", set(df1.columns) == set(df2.columns))

# Combine
df = pd.concat([df1, df2], ignore_index=True)

print("Combined dataset:", df.shape)
print(df.head())
```

```python
print("Total records:", len(df))
print("Duplicate records:", df.duplicated().sum())
print("Missing values:", df.isnull().sum().sum())

print("\nTarget distribution:")
print(df["Actual_Delay_Days"].describe())

print("\nDelivery status:")
print(df["Actual_Delivery_Status"].value_counts())
```

```python
df.to_csv(
    "../data/processed/combined_supplier_delivery_data.csv",
    index=False
)

print("Combined dataset saved!")
```

```python
import pandas as pd
import numpy as np

df = pd.read_csv("../data/processed/combined_supplier_delivery_data.csv")

print("Dataset Shape:", df.shape)
print("\nFirst 5 rows:")
display(df.head())

print("\nData Types:")
display(df.dtypes)

print("\nMissing Values:")
print(df.isnull().sum().sum())

print("\nDuplicate Rows:")
print(df.duplicated().sum())

print("\nUnique Suppliers:", df["Supplier_ID"].nunique())
print("Unique Items:", df["Item_ID"].nunique())
```

```python
print(df["Actual_Delay_Days"].describe())

print("\nDelivery Status:")
print(df["Actual_Delivery_Status"].value_counts())

print("\nDelay Distribution:")
display(df["Actual_Delay_Days"].value_counts().sort_index())
```

```python
df["Promised_Date"] = pd.to_datetime(df["Promised_Date"], errors="coerce")
df["Actual_Delivery_Date"] = pd.to_datetime(
    df["Actual_Delivery_Date"], errors="coerce"
)

df["Calculated_Delay"] = (
    df["Actual_Delivery_Date"] - df["Promised_Date"]
).dt.days

mismatches = df[
    df["Calculated_Delay"] != df["Actual_Delay_Days"]
]

print("Date mismatches:", len(mismatches))
```

```python
supplier_analysis = (
    df.groupby("Supplier_ID")
    .agg(
        Total_Orders=("Supplier_ID", "count"),
        Average_Delay=("Actual_Delay_Days", "mean"),
        On_Time_Rate=("Actual_Delivery_Status",
                      lambda x: (x == "On Time").mean() * 100)
    )
    .sort_values("Total_Orders", ascending=False)
)

display(supplier_analysis.head(20))
```
